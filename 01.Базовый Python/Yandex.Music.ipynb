**Task 20**

You have seen how grouping by city and by day of the week works. Now write a function that combines these two calculations.

Create a function `number_tracks()` that will count listenings for a given day and city. It will need two parameters:

* day of the week,
* name of the city.

In the function, store in a variable the rows of the original table where the value:

* in the `day` column equals the `day` parameter,
* in the `city` column equals the `city` parameter.

To do this, apply sequential filtering with boolean indexing (or complex logical expressions in one line, if you are already familiar with them).

Then count the values in the `user_id` column of the resulting table. Save the result in a new variable. Return this variable from the function.

```python
def number_tracks(day, city):
    track_list = df[df['day']==day] 
    track_list=track_list[track_list['city']==city]
    track_list_count =track_list['user_id'].count()
    return track_list_count
```

**Задание 24**

Create a function `genre_weekday()` with four parameters:

* table (dataframe) with data,
* day of the week,
* start time stamp in 'hh:mm' format,
* end time stamp in 'hh:mm' format.

The function should return information about the top-10 genres of tracks listened to on the specified day, during the interval between the two time stamps.

```python
def genre_weekday(df, day, time1, time2):
    genre_df =df[df['day']==day] 
    genre_df = genre_df[genre_df['time']<time2]
    genre_df = genre_df[genre_df['time']>time1]
    genre_df_grouped = genre_df.groupby('genre')['genre'].count()
    genre_df_sorted =genre_df_grouped.sort_values(ascending=False)
    return genre_df_sorted[:10]
```

**Task 25**

Compare the results of the `genre_weekday()` function for Moscow and St. Petersburg on Monday morning (from 7:00 to 11:00) and on Friday evening (from 17:00 to 23:00):

```python
genre_weekday(moscow_general, 'Monday', '07:00', '11:00')
```

```python
genre_weekday(spb_general, 'Monday', '07:00', '11:00')
```

```python
genre_weekday(moscow_general, 'Friday', '17:00', '23:00')
```

```python
genre_weekday(spb_general, 'Friday', '17:00', '23:00')
```

**Conclusions**

If we compare the top-10 genres on Monday morning, we can draw the following conclusions:

1. People in Moscow and St. Petersburg listen to similar music. The only difference is that the Moscow ranking includes the "world" genre, while St. Petersburg's includes jazz and classical.

2. In Moscow, there were so many missing values that the value `'unknown'` took tenth place among the most popular genres. This means that missing values occupy a significant portion of the data and threaten the validity of the study.

Friday evening does not change this picture. Some genres move up a bit, others down, but overall the top-10 remains the same.

* Thus, the second hypothesis was only partially confirmed:
* Users listen to similar music at the beginning and end of the week.

The difference between Moscow and St. Petersburg is not very pronounced. In Moscow, Russian popular music is heard more often; in St. Petersburg, jazz.

However, missing values in the data cast doubt on this result. In Moscow, there are so many that the top-10 ranking might look different if the lost genre data hadn't been lost.
